Etude préliminaire des datasets.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sleep_health_df=pd.read_csv('../data/sleep-health-and-lifestyle-dataset/Sleep_health_and_lifestyle_dataset.csv')
daily_activity_df=pd.read_csv('../data/fitbit/mturkfitbit_export_3.12.16-4.11.16/Fitabase Data 3.12.16-4.11.16/dailyActivity_merged.csv')
sleep_day_df=pd.read_csv('../data/fitbit/mturkfitbit_export_4.12.16-5.12.16/Fitabase Data 4.12.16-5.12.16/sleepDay_merged.csv')
```

```python
sleep_health_df.isnull().sum().sort_values(ascending=False)
```

```python
print(sleep_health_df["Sleep Disorder"].value_counts(dropna=False))
```

La target pourrait être sleep disorder, les valeurs NaN sont les personnes sans problèmes la nuit
Baseline potentielle: L'individu a un disorder non diagnostiqué.

```python
# Séparation Blood Pressure (string --> int)
sleep_health_df[['Systolic_BP', 'Diastolic_BP']] = sleep_health_df['Blood Pressure'].str.split('/', expand=True).astype(int)
sleep_health_df = sleep_health_df.drop(columns=['Blood Pressure', 'Person ID'])
num_cols = ['Age', 'Sleep Duration', 'Quality of Sleep', 'Physical Activity Level',
            'Stress Level', 'Heart Rate', 'Daily Steps', 'Systolic_BP', 'Diastolic_BP']
cat_cols = ['Gender', 'Occupation', 'BMI Category']
target = 'Sleep Disorder'

# Les NaN de Sleep Disorder correspondent à des personnes sans trouble du sommeil, on les remplace par "Normal"
sleep_health_df['Sleep Disorder'] = sleep_health_df['Sleep Disorder'].fillna("Normal")

# Erreur de saisie : "Normal Weight" au lieu de "Normal" redondant à la même catégorie, on corrige
sleep_health_df['BMI Category'] = sleep_health_df['BMI Category'].replace('Normal Weight', 'Normal')
```

```python
#Boxplots + Stripplot
for col in num_cols:
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Boxplot
    sns.boxplot(data=sleep_health_df, x=target, y=col, ax=axes[0], palette="Set2")
    axes[0].set_title(f'{col} par Sleep Disorder (Boxplot)')
    
    # Violin + Stripplot pour voir la distribution
    sns.violinplot(data=sleep_health_df, x=target, y=col, ax=axes[1], palette="Set2", inner=None)
    sns.stripplot(data=sleep_health_df, x=target, y=col, ax=axes[1], color='black', alpha=0.4, size=3)
    axes[1].set_title(f'{col} par Sleep Disorder (Distribution)')
    
    plt.tight_layout()
    plt.show()
```

```python
for col in cat_cols:
    plt.figure(figsize=(10, 5))
    
    # Countplot avec hue
    sns.countplot(data=sleep_health_df, x=col, hue=target, palette="Set2")
    plt.title(f'Distribution de {col} selon Sleep Disorder')
    plt.xticks(rotation=45)
    plt.legend(title='Sleep Disorder')
    plt.show()
    
    # Tableau croisé (pourcentages)
    crosstab = pd.crosstab(sleep_health_df[col], sleep_health_df[target], normalize='index') * 100
    print(f"\n=== {col} vs Sleep Disorder (%) ===")
    print(crosstab.round(2))
    print("-"*50)
```

```python
# Heatmap de corrélation
plt.figure(figsize=(12, 9))
corr = sleep_health_df[num_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title("Matrice de corrélation - Variables numériques")
plt.show()

# Pairplot (peut être long si le dataset est gros)
sns.pairplot(sleep_health_df[num_cols + [target]], hue=target, diag_kind='kde', palette="Set2", corner=True)
plt.show()
```
